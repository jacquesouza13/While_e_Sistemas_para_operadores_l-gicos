```python
# como funciona o while?
# while condicao:

# exemplo 1 - contar até onde o usuario pedir
limite = int(input("informe um numero limite"))
inicio = 0

while inicio <= limite:
  print(inicio)
  inicio = inicio + 1
```

```python
# controle com while
senha_correta = "abc123"
tentativas = 0

while tentativas < 3:
  senha = input("digite sua senha")
  if senha_correta == senha:
    print("Bem vindo ao sistema")
    break # quebrar
  else:
    print("senha incorreta")
    tentativas = tentativas + 1 # += 1
    print(f"numero de tentativas = {tentativas}")
    if tentativas == 3:
      print("senha bloqueada")
```

```python
# while True

while True:
  comando = input("para sair, digite SAIR")
  if comando.lower() == "sair":
    print('até mais')
    break
  else:
    print("continuando o sistema")
```

```python
# atividade 1 - Facil
# quero um sistema que pergunte ao usuario se quer salvar
# uma informação em uma lista de carros
# a lista - - > carros =["uno", "corola", "fusca"]
# se usuario quer colocar um novo carro, vai adicionar, se ele
# digitar sair, finaliza o sistema



carros = ["uno", "corola", "fusca"]
print ("=========Bem vindo ao sistema de cadastro de carros=========")
while True:
  novo_carro = input("Informe o nome do carro ou digite SAIR para finalizar")
  if novo_carro.lower() == "sair":
    print ("Até mais")
    break
  else:
    carros.append(novo_carro)
    for i in carros:
      print(i)

```
